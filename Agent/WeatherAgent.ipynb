```python
from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace

repo_id = "Qwen/Qwen2.5-7B-Instruct"
model = HuggingFaceEndpoint(repo_id=repo_id, temperature=0)
llm = ChatHuggingFace(llm=model)
```

```python
from langchain_community.tools import DuckDuckGoSearchRun

search_tool = DuckDuckGoSearchRun()
```

```python
import requests
from langchain_core.tools import tool
```

```python
@tool
def get_weather_data(city: str) -> str:
  "This fetches live weather of a given place"

  api_key = "521ab6f873609a43bd1ecd7ada6134c0"

  url = f"http://api.weatherstack.com/current?access_key={api_key}&query={city}"

  response = requests.get(url)

  return response.json()

```

```python
!pip install langchain langgraph
```

```python
from langchain.agents import create_agent
from langchain_core.callbacks.stdout import StdOutCallbackHandler

```

```python
agent = create_agent(
    model=llm,
    tools=[search_tool, get_weather_data],
)
```

```python

response = agent.invoke({
    "messages": [("user", "Find the capital of Nepal, then find the it's weather condition?")]
})

print(response["messages"][-1].content)
```

```python
response = agent.invoke({
    "messages": [
        ("user", "Find the capital of Nepal."),
        ("user", "Now fetch the current detailed weather for Jhapa, Nepal.")
    ]
})
print(response["messages"][-1].content)
```
