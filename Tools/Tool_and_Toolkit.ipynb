```python
#Custom Tool Using tool decorator
```

```python
from langchain_core.tools import tool
```

```python
#creating function

def addition(a,b):
  """Add two numbers"""
  return a+b
```

```python
# adding type hints

def addition(a: int,b: int) -> int:
  """Add two numbers"""
  return a+b
```

```python
# add tool decorator

@tool
def addition(a: int,b: int) -> int:
  """Add two numbers"""
  return a+b
```

```python
result = addition.invoke({"a":4, "b":6})

print(result)
```

```python
print(addition.name)
print(addition.description)
print(addition.args)
```

```python
print(addition.args_schema.model_json_schema())
```

```python
# Custom tool using StructuredTool and Pydantic below
```

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel,Field
```

```python
class AdditionInput(BaseModel):
    a: int = Field(description="First number")
    b: int = Field(description="Second number")
```

```python
def addition_func(a: int,b: int) -> int:
  return a+b
```

```python
addition_tool = StructuredTool.from_function(
    func=addition_func,
    name="addition",
    description="Add two numbers",
    args_schema=AdditionInput
)
```

```python
result = addition_tool.invoke({"a":4,"b":3})

print(result)
```

```python
print(addition_tool.name)
print(addition_tool.description)
print(addition_tool.args)
```

```python
# custom tool using BaseTool . All other tools like like @tool, StructuredTool are built on top of BaseTool
```

```python
from langchain.tools import BaseTool
from typing import Type
```

```python
class AdditionInput(BaseModel):
    a: int = Field(description="First number")
    b: int = Field(description="Second number")
```

```python
class MultiplyTool(BaseTool):
  name: str = "addition"
  description: str = "Add two numbers"
  args_schema: Type[BaseModel] = AdditionInput

  def _run(self, a: int, b: int) -> int:
    return a+b
```

```python
multiply_tool = MultiplyTool()
```

```python
result = multiply_tool.invoke({"a":3,"b":3})

print(result)
```

```python
# Toolkit: Collection of related Tools
```

```python
from langchain_core.tools import tool

@tool
def addition(a: int,b: int) -> int:
  """Add two numbers"""
  return a+b

@tool
def multiply(a: int,b: int) -> int:
  """Multiply two numbers"""
  return a*b

```

```python
class MathToolKit:
  def get_tools(self):
    return [addition,multiply]
```

```python
toolkit = MathToolKit()

tools = toolkit.get_tools()

for tool in tools:
  print(tool.name)
  print(tool.description)
```
